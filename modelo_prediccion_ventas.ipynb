```python
import pandas as pd
from sklearn.linear_model import LinearRegression

# 1. Datos históricos: (Inversión en publicidad $, Ventas logradas $)
datos_marketing = [
    (100, 1100),
    (200, 2050),
    (300, 3100),
    (400, 3900),
    (500, 4950)
]

df_ml = pd.DataFrame(datos_marketing, columns=['Publicidad', 'Ventas'])

# 2. Separamos la variable que usamos para predecir (X) y la que queremos adivinar (y)
X = df_ml[['Publicidad']]
y = df_ml['Ventas']

# 3. Creamos y entrenamos el modelo de Inteligencia Artificial (.fit)
modelo = LinearRegression()
modelo.fit(X, y)

print("¡Modelo entrenado con éxito! Listo para hacer predicciones.")
```

```python
# Creamos una tabla con los nuevos montos que queremos evaluar invertir
nueva_inversion = pd.DataFrame({'Publicidad': [650, 800]})

# El modelo calcula cuánto venderemos con cada inversión
predicciones = modelo.predict(nueva_inversion)

print("Si invertimos $650 en publicidad, las ventas estimadas son: $", round(predicciones[0], 2))
print("Si invertimos $800 en publicidad, las ventas estimadas son: $", round(predicciones[1], 2))
```

```python
import matplotlib.pyplot as plt

# 1. Dibujamos los datos históricos reales (puntos azules)
plt.scatter(df_ml['Publicidad'], df_ml['Ventas'], color='blue', label='Datos Históricos')

# 2. Dibujamos la línea que aprendió el modelo (línea roja)
plt.plot(df_ml['Publicidad'], modelo.predict(X), color='red', label='Línea de Tendencia (Modelo)')

# 3. Dibujamos nuestras 2 predicciones futuras (estrellas verdes)
plt.scatter(nueva_inversion['Publicidad'], predicciones, color='green', marker='*', s=200, label='Predicciones Futuras ($650 y $800)')

plt.title('Predicción de Ventas según Inversión en Publicidad')
plt.xlabel('Inversión en Publicidad ($)')
plt.ylabel('Ventas ($)')
plt.legend()
plt.grid(True, linestyle='--', alpha=0.6)
plt.show()
```
